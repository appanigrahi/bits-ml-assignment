## 1. Import Required Libraries

```python
# Data Handling
import pandas as pd
import numpy as np
!pip install scikit-learn

# Dataset
from sklearn.datasets import load_breast_cancer

# Data Split
from sklearn.model_selection import train_test_split

# Machine Learning Models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier

# Evaluation Metrics
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report
)

# Save Models
import joblib

print("All libraries imported successfully")
```

## 2.Load the Breast Cancer Wisconsin Dataset
### We have the Breast Cancer Wisconsin dataset loaded into a DataFrame. This dataset contains 569 records and 30 features, satisfying the assignment requirements


```python
# Load Breast Cancer Wisconsin Dataset

data = load_breast_cancer()

# Create DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)

# Add Target Column
df["target"] = data.target

# Display dataset information
print("Dataset Shape:", df.shape)
print("\nTarget Values:")
print(df["target"].value_counts())

# Display first 5 records
df.head()
```

## 3.Split Dataset into Features, Target, Training Data and Test Data
### we must divide the dataset into:
Training Data (80%) → Used to train the ML models.  
Testing Data (20%) → Used to calculate Accuracy, AUC, Precision, Recall, F1, and MCC metrics

```python
# Features (X) and Target (y)

X = df.drop("target", axis=1)
y = df["target"]

# Split data into Training and Testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)
```

## 4.Train Logistic Regression Model

```python
# Logistic Regression Model

lr_model = LogisticRegression(max_iter=10000)

# Train the model
lr_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully")
```

## 5.Calculate All Required Metrics for Logistic Regression
### The assignment requires these metrics for every model:
Accuracy  
AUC Score    
Precision   
Recall  
F1 Score   
MCC Score

```python
# Make Predictions

y_pred = lr_model.predict(X_test)

# Probability predictions (needed for AUC)
y_prob = lr_model.predict_proba(X_test)[:, 1]

# Calculate Metrics

accuracy = accuracy_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_prob)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
mcc = matthews_corrcoef(y_test, y_pred)

# Display Results

print("Logistic Regression Results")
print("-" * 40)
print("Accuracy :", round(accuracy, 4))
print("AUC      :", round(auc, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))
print("MCC Score:", round(mcc, 4))
```

## 6.Save the Logistic Regression Model

```python
# Save Logistic Regression Model

joblib.dump(lr_model, "logistic_regression.pkl")

print("Model saved successfully")
```

## 7.Train the Decision Tree Classifier

```python
# Decision Tree Classifier

dt_model = DecisionTreeClassifier(
    random_state=42
)

# Train the model
dt_model.fit(X_train, y_train)

print("Decision Tree model trained successfully")
```

## 8.Calculate All Required Metrics for Decision Tree
### The assignment requires these metrics for every model:
Accuracy  
AUC Score  
Precision  
Recall  
F1 Score  
MCC Score  

```python
# Decision Tree Predictions

dt_pred = dt_model.predict(X_test)

# Probability predictions for AUC
dt_prob = dt_model.predict_proba(X_test)[:, 1]

# Metrics

dt_accuracy = accuracy_score(y_test, dt_pred)
dt_auc = roc_auc_score(y_test, dt_prob)
dt_precision = precision_score(y_test, dt_pred)
dt_recall = recall_score(y_test, dt_pred)
dt_f1 = f1_score(y_test, dt_pred)
dt_mcc = matthews_corrcoef(y_test, dt_pred)

# Display Results

print("Decision Tree Results")
print("-" * 40)
print("Accuracy :", round(dt_accuracy, 4))
print("AUC      :", round(dt_auc, 4))
print("Precision:", round(dt_precision, 4))
print("Recall   :", round(dt_recall, 4))
print("F1 Score :", round(dt_f1, 4))
print("MCC Score:", round(dt_mcc, 4))
```

## 9.Save the Decision Tree Model

```python
# Save Decision Tree Model

joblib.dump(dt_model, "decision_tree.pkl")

print("Decision Tree model saved successfully")
```
